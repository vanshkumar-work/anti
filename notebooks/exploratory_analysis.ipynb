### 1. Dataset Ingestion and Structural Overview

```python
df_raw = load_raw_dataset('../data/raw/deceptive_opinion.csv')
print('Dataset dimensions:', df_raw.shape)
df_raw.head()
```

### 2. Class Balance and Polarity Verification

```python
print('Class balance:\n', df_raw['deceptive'].value_counts())
print('\nPolarity balance:\n', df_raw['polarity'].value_counts())
print('\nUnique hotels represented:', df_raw['hotel'].nunique())
```

### 3. Review Length and Quantitative Diagnostics

```python
df_raw['word_count'] = df_raw['text'].apply(lambda x: len(x.split()))
df_raw['char_count'] = df_raw['text'].apply(len)

truthful_words = df_raw[df_raw['label'] == 0]['word_count']
deceptive_words = df_raw[df_raw['label'] == 1]['word_count']

print(f'Truthful Review Word Count:  Mean={truthful_words.mean():.2f}, Median={truthful_words.median():.2f}')
print(f'Deceptive Review Word Count: Mean={deceptive_words.mean():.2f}, Median={deceptive_words.median():.2f}')
```

### 4. N-Gram Vocabulary Extraction (TF-IDF Unigrams + Bigrams)

```python
vectorizer = build_tfidf_vectorizer(ngram_range=(1, 2), min_df=3, max_features=5000)
X_tfidf = vectorizer.fit_transform(df_raw['text'])
feature_names = vectorizer.get_feature_names_out()
print('Extracted vocabulary size:', len(feature_names))
print('Sample vocabulary n-grams:', feature_names[100:115])
```

### 5. Key Findings and Linguistic Hypotheses
1. **Spatial Detail:** Truthful reviews mention specific spatial references ('bathroom', 'street', 'floor', 'location', 'elevator', 'valet').
2. **Emotional Hyperbole:** Deceptive reviews contain high frequencies of abstract superlatives ('perfection', 'breathtaking', 'heavenly', 'pampered').
3. **First-Person Framing:** Deceptive texts more frequently frame reviews around personal narratives ('my husband and i', 'our anniversary', 'we will return').
